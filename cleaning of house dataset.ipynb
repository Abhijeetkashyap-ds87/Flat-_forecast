```python
h['noOfFloor']=h['noOfFloor'].str.strip(' ').str.split('F').str.get(0)
```

```python
def change_price(l):
    if type(l)==float:# for NaN values
        return l
    elif len(l)==2:
        if l[1]=='Lac':
            return round(float(l[0])*0.01,2)
        elif l[1]=='Crore':
            return round(float(l[0]),2)
    else:
        return ''.join(l)
```

```python
h['price']=h['price'].str.split(' ').apply(change_price)
```

```python
def remove_comma(x):
    if type(x)==float:# for nan values
        return x
    else:
        l=x.split(',')
        return float(''.join(l))
```
