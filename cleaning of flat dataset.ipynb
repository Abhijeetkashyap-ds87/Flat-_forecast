```python
# converting to crore
def change_price(l):
    if type(l)==float:# for NaN values
        return l
    elif len(l)==2:
        if l[1]=='Lac':
            return round(float(l[0])*0.01,2)
        elif l[1]=='Crore':
            return round(float(l[0]),2)
    else:
        return ''.join(l)
```

```python
f_dc['price']=f_dc['price'].str.split(' ').apply(change_price)
```

```python
l=f_dc['area_persqft'].str.split(',')[0]
float(''.join(l))
```

```python
def remove_comma(x):
    if type(x)==float:# for nan values
        return x
    else:
        l=x.split(',')
        return float(''.join(l))
```

```python
f_dc['area_persqft']=f_dc['area_persqft'].apply(remove_comma)
```

```python
f_dc.sample(5)
```

```python
def create(price,area_persqft):
    if isinstance(price,float):
        return round((price*10000000)/area_persqft,3)
    else:
        return np.nan
```

```python
f_dc['area'] = f_dc.apply(lambda row: create(row['price'], row['area_persqft']), axis=1)
```
